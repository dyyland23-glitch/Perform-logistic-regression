### Produce a confusion matrix

Data professionals often like to know the types of errors made by an algorithm. To obtain this information, produce a confusion matrix.

```python
### YOUR CODE HERE ###
import sklearn.metrics as metrics
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 1. Compute the raw confusion matrix
cm = metrics.confusion_matrix(y_test, y_pred, labels=clf.classes_)

# 2. Define text labels for each quadrant
quadrant_names = ["True Negative", "False Positive", "False Negative", "True Positive"]
counts = [f"{val:,}" for val in cm.flatten()]
percentages = [f"{val:.2%}" for val in cm.flatten() / np.sum(cm)]

# Combine label, raw count, and percentage
formatted_labels = [f"{n}\n{c}\n({p})" for n, c, p in zip(quadrant_names, counts, percentages)]
formatted_labels = np.asarray(formatted_labels).reshape(2, 2)

# 3. Plot using seaborn heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=formatted_labels,
    fmt="",
    cmap="Blues",
    cbar=False,
    xticklabels=["Dissatisfied (0.0)", "Satisfied (1.0)"],
    yticklabels=["Dissatisfied (0.0)", "Satisfied (1.0)"]
)
plt.xlabel("Predicted label", fontsize=11)
plt.ylabel("True label", fontsize=11)
plt.title("Confusion Matrix with Quadrant Labels", fontsize=12)
plt.show()
```

```python
### YOUR CODE HERE ###


import sklearn.metrics as metrics



# Calculate the confusion matrix
cm = metrics.confusion_matrix(y_test, y_pred, labels=clf.classes_)

# Display the confusion matrix
disp = metrics.ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=clf.classes_)
disp.plot()
plt.show()


```

## What stands out to you about the confusion matrix?

The model produces a relatively balanced number of true positives and true negatives, but there is a noticeable disparity in the types of errors: it predicts substantially more false negatives (predicting dissatisfied when the passenger was actually satisfied) than false positives (predicting satisfied when dissatisfied).

## Did you notice any difference in the number of false positives or false negatives that the model produced?

Yes. False negatives are notably higher than false positives because passengers rating in-flight entertainment around moderate scores (like 3) might still report being satisfied overall due to other aspects of the trip (seat comfort, service, delays) that this single-variable model cannot see.
